# Feature Engineering

```python
df["readmission_30_days"] = np.where(
    df["readmitted"] == "<30",1,0)
```

```python
df["total_prior_visits"] = (
    df["number_inpatient"] +
    df["number_outpatient"] +
    df["number_emergency"])
```

```python
df["long_stay"] = np.where(
    df["time_in_hospital"] >= 7,1,0)
```

```python
df["polypharmacy"] = np.where(
    df["num_medications"] >= 10,1,0)
```

```python
df["high_utilizer"] = np.where(
    df["total_prior_visits"] >= 5,1,0)
```
